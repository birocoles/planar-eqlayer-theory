```python
import numpy as np
```

```python
def spherical2Cartesian(intensity, inclination, declination):
    inc = np.deg2rad(inclination)
    dec = np.deg2rad(declination)
    vector = intensity*(
        np.array([np.cos(inc)*np.cos(dec), np.cos(inc)*np.sin(dec), np.sin(inc)])
    )
    return vector
```

```python
k = np.array([3j, 4j, 5])
```

```python
k
```

```python
t = spherical2Cartesian(1., 10., 23.7)
```

```python
h = spherical2Cartesian(3., 45, -10.3)
```

```python
v = spherical2Cartesian(5., -31.3, 18)
```

```python
alpha_d = np.sum(t*k)*np.sum(h*k)
```

```python
alpha_a = np.sum(t*k)*np.sum(v*k)
```

```python
ratio_d_a = alpha_d / alpha_a 
```
