# Ambiguity in magnetization direction of a dipole

```python
import numpy as np
```

```python
def vector_a(p, q):
    xp, yp, zp = p
    xq, yq, zq = q
    
    r = np.sqrt((xp - xq)**2 + (yp - yq)**2 + (zp - zq)**2)
    Dx = -(xp - xq)/r**3
    Dy = -(yp - yq)/r**3
    Dz = -(zp - zq)/r**3
    
    a = np.array([Dx, Dy, Dz])

    return a
```

```python
def vector_u(I, D):
    inc = np.rad2deg(I)
    dec = np.rad2deg(D)
    
    cosI = np.cos(inc)
    sinI = np.sin(inc)
    cosD = np.cos(dec)
    sinD = np.sin(dec)
    
    ux = cosI*cosD
    uy = cosI*sinD
    uz = sinI

    u = np.array([ux, uy, uz])
    
    return u
```

```python
p = [2, 3, 0]
q = [3.4, 4, 5]

a = vector_a(p, q)
u1 = vector_u(-10, 30)
u2 = vector_u(20, 45)
```

```python
norm_a  = np.sqrt(np.sum(a*a))
norm_u1 = np.sqrt(np.sum(u1*u1))
norm_u2 = np.sqrt(np.sum(u2*u2))
```

```python
np.allclose(norm_u1, 1)
```

```python
np.allclose(norm_u2, 1)
```

```python
cos_a_u1 = np.dot(a, u1)/norm_a
cos_a_u2 = np.dot(a, u2)/norm_a
```

```python
V1 = -np.dot(a, u1)
V2 = -np.dot(a, u2)
```

```python
np.allclose(V1, -norm_a*cos_a_u1)
```

```python
np.allclose(V2, -norm_a*cos_a_u2)
```

```python
np.allclose(V1, V2*cos_a_u1/cos_a_u2)
```
