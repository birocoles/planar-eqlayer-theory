# General function for epsilon

```python
import numpy as np
from scipy.integrate import quad
from matplotlib import pyplot as plt
```

```python
from scipy.special import gammainc, gamma
from numpy.testing import assert_almost_equal
```

```python
# Function for evaluating the integrand
def integrand(k, Dz, orders):
    aux = np.sum(orders)
    return (k**(2*aux - 1))*np.exp(-2*Dz*k)
```

```python
# cut-off wavenumber and distance from continuatioj plane and top of the sources
k_epsilon = 0.01
Dz = 100.
```

```python
derivative_orders = (3, 3, 3)
```

```python
# result obtained via numerical integration
result, error = quad(integrand, k_epsilon, np.inf, args=(Dz, derivative_orders))
```

```python
result, error
```

```python
derivative_orders = (0, 0, 0)
```

```python
# result obtained via numerical integration
result, error = quad(integrand, k_epsilon, np.inf, args=(Dz, derivative_orders))
```

```python
result, error
```

```python
# result obtained via scipy.special
a = 1e-7
result2 = gamma(a)*(1. - gammainc(a, 2*Dz*k_epsilon))
```

```python
# comparison
assert_almost_equal(result, result2, decimal=8)
```
