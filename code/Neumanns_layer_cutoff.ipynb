### Estimate the pseudogravity anomaly on the layer plane

```python
# set a layer that crosses the sources
z_layer = 200
```

```python
print('The distance from coordinates grid to layer is {} m'.format(z_layer-coordinates_grid['z']))
```

```python
# compute the SEDM
R2 = inverse_distance.sedm_BTTB(
    data_grid = coordinates_grid, 
    delta_z = z_layer - coordinates_grid['z'],
    ordering = 'xy'
)
```

```python
# compute the vertical derivative of inverse distance function
Grad = inverse_distance.grad_BTTB(
    data_grid = coordinates_grid,
    delta_z = z_layer - coordinates_grid['z'],
    SEDM = R2,
    ordering = 'xy',
    components=['z']
)
```

```python
# compute the eigenvalues matrix
L_Gz = convolve.eigenvalues_BCCB(BTTB_metadata=Grad['z'])
```

```python
# estimate the pseudogravity anomaly on the layer plane
pseudo_grav_eql_convergence, pseudo_grav_layer = eqlayer.method_iterative_deconvolution_TOB20(
    eigenvalues_matrices=[L_Gz],
    data_vectors=[pseudo_grav],
    epsilon=1e-7,
    ITMAX=100
)
```

```python
# Define tentative k_esilon values
k_epsilons = np.linspace(1e-2, 6e-2, 50)
```

```python
k_epsilons[0], k_epsilons[-1]
```

```python
# compute the tensor components (second derivatives) of the inverse distance function
Tensor = inverse_distance.grad_tensor_BTTB(
    data_grid = coordinates_grid,
    delta_z = z_layer - coordinates_grid['z'],
    SEDM = R2,
    ordering = 'xy',
    components=['xx', 'xy', 'xz', 'yy', 'yz', 'zz']
)
```

```python
# compute the eigenvalues matrices associated to second derivatives of inverse distance function
L_Tensor = dict()
for component in ['xx', 'xy', 'xz', 'yy', 'yz', 'zz']:
    L_Tensor[component] = convolve.eigenvalues_BCCB(BTTB_metadata=Tensor[component])
```

```python
# directional factor to compute the approximated total-field anomaly
a = utils.directional_factors(t=h_hat, u=F_hat)
```

```python
# Run this cell may take some time.

start = time.time()

# define the approximated total-field anomaly fit versus truncating wavenumber
data_fits = []
for k_epsilon in k_epsilons:
    # define the low-pass filter
    lowpass = filters.cuttof_frequency(wavenumbers = wavenumbers, max_freq = k_epsilon)
    # apply the cutoff associated with the current k_epsilon
    FT_pseudo_grav_layer_cutoff = convolve.compute(FT_pseudo_grav_layer, [lowpass])
    # transform the field to space domain
    pseudo_grav_layer_cutoff = transforms.IDFT(FT_pseudo_grav_layer_cutoff, unpad=True)
    # approximated total-field anomaly
    approx_tfa_predicted = np.zeros_like(bz)
    for component in ['xx', 'xy', 'xz', 'yy', 'yz']:
        approx_tfa_predicted[:] += convolve.product_BCCB_vector(
            eigenvalues = a[component]*L_Tensor[component],
            ordering = 'row',
            v = data_structures.grid_xy_full_matrix_to_flatten(
                grid = pseudo_grav_layer_cutoff, ordering = 'xy'
            )
        )
    approx_tfa_predicted[:] *= constants.MAGNETIC_CONST
    # compute the data fit
    residuals = approx_tfa_predicted - approx_tfa
    data_fits.append(np.sum(residuals*residuals)/residuals.size)

end = time.time()

print('elapsed time = {:.3f} s'.format(end - start))
```

```python
# generic error bound
generic_error_bound = []
for k_epsilon in k_epsilons:
    result, error = quad(integrand, k_epsilon, np.inf, args=(z_layer-coordinates_grid['z'], 2))
    generic_error_bound.append((2*np.pi*total_moment*total_moment)*result)
generic_error_bound = np.asarray(generic_error_bound)
```

```python
# k_epsilon leading to the optimum data fit
k_epsilon_optimum = k_epsilons[np.argmin(data_fits)]
```

```python
print('Truncating wavenumber leading to minimum error: {:.5f}'.format(k_epsilon_optimum))
```

```python
# theoretical error bound and minimum computed error
result, error = quad(integrand, k_epsilon_optimum, np.inf, args=(z_layer-coordinates_grid['z'], 2))
generic_error_bound_optimum = (2*np.pi*total_moment*total_moment)*result

print(generic_error_bound_optimum, np.min(data_fits))
```

### Compute the data fit

```python
# list for storing the predicted magnetic field components
predicted_fields = dict()
```

```python
# magnetic induction field component x
result = np.zeros_like(bx)

for h_component, component in zip(h_hat, ['xx', 'xy', 'xz']):
    result[:] += h_component*convolve.product_BCCB_vector(
        eigenvalues = L_Tensor[component],
        ordering = 'row',
        v = data_structures.grid_xy_full_matrix_to_flatten(grid=pseudo_grav_layer_cutoff, ordering='xy')
    )

predicted_fields['bx'] = constants.MAGNETIC_CONST*result

# magnetic induction field component y
result = np.zeros_like(by)

for h_component, component in zip(h_hat, ['xy', 'yy', 'yz']):
    result[:] += h_component*convolve.product_BCCB_vector(
        eigenvalues = L_Tensor[component],
        ordering = 'row',
        v = data_structures.grid_xy_full_matrix_to_flatten(grid=pseudo_grav_layer_cutoff, ordering='xy')
    )

predicted_fields['by'] = constants.MAGNETIC_CONST*result

# magnetic induction field component z
result = np.zeros_like(bz)

for h_component, component in zip(h_hat, ['xz', 'yz', 'zz']):
    result[:] += h_component*convolve.product_BCCB_vector(
        eigenvalues = L_Tensor[component],
        ordering = 'row',
        v = data_structures.grid_xy_full_matrix_to_flatten(grid=pseudo_grav_layer_cutoff, ordering='xy')
    )

predicted_fields['bz'] = constants.MAGNETIC_CONST*result
```

```python
# amplitude of anomalous B field
predicted_fields['amplitude'] = np.sqrt(
    predicted_fields['bx']**2 + 
    predicted_fields['by']**2 + 
    predicted_fields['bz']**2
)
```

```python
# reduction to the pole
result = convolve.product_BCCB_vector(
    eigenvalues = L_Tensor['zz'],
    ordering = 'row',
    v = data_structures.grid_xy_full_matrix_to_flatten(grid=pseudo_grav_layer_cutoff, ordering='xy')
)

predicted_fields['rtp'] = constants.MAGNETIC_CONST*result
```

```python
# approximated total-field anomaly
result = np.zeros_like(bz)

for component in ['xx', 'xy', 'xz', 'yy', 'yz']:
    result[:] += convolve.product_BCCB_vector(
        eigenvalues = a[component]*L_Tensor[component],
        ordering = 'row',
        v = data_structures.grid_xy_full_matrix_to_flatten(grid=pseudo_grav_layer_cutoff, ordering='xy')
    )

predicted_fields['atfa'] = constants.MAGNETIC_CONST*result
```
