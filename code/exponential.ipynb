```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def bound_spectrum(k, Omega, Dz):
    return Omega*np.exp(k*Dz)/k
```

```python
# sample interval in space domain
dx = 10.

# maximum frequency
f_max = 1/(2*dx)
```

```python
# maximum wavenumber
k_max = 2*np.pi*f_max
```

```python
k_max
```

```python
k = np.linspace(1e-5, k_max, 500)
```

```python
# integral depending only on the sources
Omega = 1
```

```python
# difference between z of the layer and of the sources top
Dz = 300
```

```python
k_ast = 1/Dz
```

```python
bound = bound_spectrum(k, Omega, Dz)
bound_max = np.max(bound)
bound_min = np.min(bound)
```

```python
plt.figure(figsize=(10,4))

plt.subplot(121)
plt.plot(k, bound, 'k-')
plt.plot([k_ast, k_ast], [0, bound_max], 'r--')
plt.grid()

plt.subplot(122)
plt.plot(k, bound, 'k-')
plt.plot([k_ast, k_ast], [0, bound_max], 'r--')
plt.xlim(k_ast-0.005, k_ast+0.005)
plt.ylim(bound_min, bound_min+400)
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(121)
plt.plot(k, bound, 'k-')
plt.plot([k_ast, k_ast], [0, bound_max], 'r--')
plt.yscale('log')
plt.grid()

plt.subplot(122)
plt.plot(k, bound, 'k-')
plt.plot([k_ast, k_ast], [0, bound_max], 'r--')
plt.xlim(k_ast-0.002, k_ast+0.002)
plt.ylim(bound_min, bound_min+400)
plt.grid()

plt.show()
```
